Install dependencies in conda environment:

```conda install <package_names>```

- eg: ```conda install pandas numpy python=3.9```

- eg: ```conda install -n myenv numpy python=3.9```

- eg: ```conda install -p path/to/env python=3.9```

Params:
- ```-n/–name``` : name of environment
- ```-p/–prefix``` : full path of environment
- ```--update-deps``` : update dependencies that have available updates
- ```--update-all/-all``` : update all dependencies
- ```-y``` : answer yes with all questions about installation 

Update dependencies in conda environment:

```conda update -n myenv <package_names>```

- ```eg: conda update -n myenv scipy```
