To remove packages in conda environment:

```conda remove <package_names>```

- eg: ```conda remove pandas```

- eg: ```conda remove -n myenv scipy curl wheel```

- eg: ```conda remove -n myenv --all```
    
- eg: ```conda remove -n myenv --all --keep-env```

Params:
- –all : remove all packages
- --keep-env : full path of environment
